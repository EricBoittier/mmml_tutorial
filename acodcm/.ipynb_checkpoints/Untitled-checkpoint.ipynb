```python
import numpy as np

# Source - https://stackoverflow.com/a/56515845
# Posted by sgt pepper
# Retrieved 2026-05-19, License - CC BY-SA 4.0

def unit_vector(vector):
    """ Returns the unit vector of the vector"""
    return vector / np.linalg.norm(vector)

def angle_between(vector1, vector2):
    """ Returns the angle in radians between given vectors"""
    v1_u = unit_vector(vector1)
    v2_u = unit_vector(vector2)
    minor = np.linalg.det(
        np.stack((v1_u[-2:], v2_u[-2:]))
    )
    if minor == 0:
        raise NotImplementedError('Too odd vectors =(')
    return np.sign(minor) * np.arccos(np.clip(np.dot(v1_u, v2_u), -1.0, 1.0))
```

```python
from itertools import combinations

h5_files = list(Path("/mmhome/boittier/home/mmml/acetone/").glob("N*/*/R*/vac_nvt_jaxmd_nvt.h5"))
h5_files
```

```python

O_idxs = np.array([6 + i*10 for i in range(12)])
```

```python
angles = []
dists2 = []

cl_idxs = np.array([0, 1])
Natoms = 10
Nmols = 12


for file in h5_files:
    plt.clf()
    try:
        print(file)
        
        hf = h5py.File(file, "r")
        R = np.array(hf["positions"])
        PE = np.array(hf["potential_energy"])
        KE = np.array(hf["kinetic_energy"])
        t = np.array(hf["time_ps"])

        Nmols = R.shape[1]//10
        R = R[KE < 10]

        for i in range(len(R)):
            atoms = R[i]
            for a,b in (list(combinations(range(Nmols), 2))):
                mono1_idx = a*Natoms + cl_idxs 
                mono2_idx = b*Natoms + cl_idxs 
            
                mono1_pos = atoms[mono1_idx]
                mono2_pos = atoms[mono2_idx]
                v1 = mono1_pos[0] - mono1_pos[1]
                v2 = mono2_pos[0] - mono2_pos[1]
                m1 =atoms[a*Natoms]
                m2 = atoms[b*Natoms]
                m_dist = np.linalg.norm(m1 - m2)
            
                # v1 = np.random.normal(size=(3))
                # v2 = np.random.normal(size=(3))
                angle = angle_between(v1, v2)
                angle = np.rad2deg(angle)
                angles.append(angle)
                dists2.append(m_dist)

    except Exception as e:
        print(e)
```

```python
import numpy as np
import matplotlib.pyplot as plt


counts, bin_edges = np.histogram(dists2, bins=200, range=(0, 7.5), density=False)

r_inner = bin_edges[:-1]
r_outer = bin_edges[1:]
r = 0.5 * (r_inner + r_outer)

shell_volumes = (4.0 / 3.0) * np.pi * (r_outer**3 - r_inner**3)

rho_c = 110
n_frames = 1
n_ref = 1
ideal_counts = n_frames * n_ref * rho_c * shell_volumes

g_r = counts / ideal_counts

plt.plot(r, g_r)
plt.xlabel("C-C distance")
plt.ylabel("g(r)")
plt.legend()
plt.show()
```
