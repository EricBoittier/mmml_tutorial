```python

import matplotlib.pyplot as plt

uuid="8a9ff4aa-0c48-4a8e-b8f4-504eae6276d7"
PARAMS = f"/mmhome/boittier/home/ckpts/ef_run3/params-{uuid}.json"
CONFIG = f"/mmhome/boittier/home/ckpts/ef_run3/config-{uuid}.json"


EF = [0.0, 0.0, 0.0]

def get_dih_scan_data(PARAMS, CONFIG, EF, plot=True):

    calc = AseCalculatorEF(
        PARAMS,
        config_path=CONFIG,
        electric_field=EF,  # optional default field (model input units)
        field_scale=100,  # E_physical [au] = E_input * field_scale
    )
    
    atoms.calc = calc
    atoms.get_potential_energy()
    atoms.get_forces()
    
    
    atoms_dih_scan_E = []
    atoms_dih_scan_F = []
    atoms_dih_scan_D = []
    
    for _ in atoms_dih_scan:
        atoms.set_positions(_.get_positions())
        atoms_dih_scan_E.append(atoms.get_potential_energy())
        atoms_dih_scan_F.append(atoms.get_forces())
        atoms_dih_scan_D.append(atoms.get_dipole_moment())
    
    atoms_dih_scan_F = np.array(atoms_dih_scan_F)
    atoms_dih_scan_D = np.array(atoms_dih_scan_D)
    
    
    if plot:
        plt.plot(phi_angles,atoms_dih_scan_E)
        plt.xlim(-180,180)
        plt.show()
        for i in range(4):
            plt.plot(phi_angles, np.linalg.norm(atoms_dih_scan_F[:,i,:], axis=1), label=atoms.get_atomic_numbers()[i], alpha=0.5, linestyle="--")
        plt.legend()
        plt.xlim(-180,180)
        plt.show()
        
        plt.plot(phi_angles, np.linalg.norm(atoms_dih_scan_D, axis=1), label=atoms.get_atomic_numbers()[i], alpha=0.5, linestyle="--")
        plt.legend()
        plt.xlim(-180,180)
        plt.show()
    
    return {"phi_angles": phi_angles, "atoms_dih_scan_E": atoms_dih_scan_E, "atoms_dih_scan_F":atoms_dih_scan_F, "atoms_dih_scan_D":atoms_dih_scan_D, "EF": EF}
```

```python
def chemcoord_to_ase(cart: cc.Cartesian):
    return ase.Atoms(cart["atom"], cart[["x", "y", "z"]])


atoms_dih_scan = []
phi_angles = np.arange(-179, 179, 5)
for phi in phi_angles:
    distortion = hper_a.copy()
    distortion.unsafe_loc[3, 'dihedral'] = phi
    distortion_xyz = distortion.to_xyz()
    
    atoms = chemcoord_to_ase(distortion_xyz)
    atoms_dih_scan.append(atoms)
```

```python
import numpy as np

# traj: ASE Trajectory, or list/tuple of ASE Atoms from traj[i]
# indices: list/array of frame indices, e.g. [0, 10, 20, 50]
out_path = "dih_frames.npz"

R_list = []
N_list = []
Z_ref = atoms_dih_scan[0].get_atomic_numbers()

for i in range(len(atoms_dih_scan)):
    atoms = atoms_dih_scan[int(i)]  # use traj[int(i)] if indices are floats
    R = np.asarray(atoms.get_positions(), dtype=np.float64)
    Z = np.asarray(atoms.get_atomic_numbers(), dtype=np.int32)
    if Z_ref is None:
        Z_ref = Z
    elif not np.array_equal(Z_ref, Z):
        raise ValueError(f"Frame {i}: atomic numbers differ from first selected frame")
    R_list.append(R)
    N_list.append(len(atoms))

R_arr = np.stack(R_list, axis=0)  # (n_frames, n_atoms, 3)
N_arr = np.asarray(N_list, dtype=np.int32)  # (n_frames,)

np.savez_compressed(out_path, R=R_arr, Z=Z_ref, N=N_arr)
print("Saved", out_path, "R", R_arr.shape, "Z", Z_ref.shape, "N", N_arr.shape)
```

```python


from tqdm import tqdm

pert = 0.01
EFs = [[0.0, 0.0, 0.0], [pert, 0.0, 0.0], [0.0, pert, 0.0], [0.0, 0.0, pert]]
all_dfs = []
for EF in EFs:
    flat_records = []
    for r in tqdm(runs):
        out_data = get_dih_scan_data(r["params"], r["config"], EF, plot=False)
    
        phi_angles = np.asarray(out_data["phi_angles"])
        energies = np.asarray(out_data["atoms_dih_scan_E"])
        forces = np.asarray(out_data["atoms_dih_scan_F"])   # (n, 4, 3)
        dipoles = np.asarray(out_data["atoms_dih_scan_D"])  # (n, 3)
        ef_vec = np.asarray(out_data["EF"])
    
        n = len(phi_angles)
    
        for i in range(n):
            row = {
                "uuid": r["uuid"],
                # "run": r["run"],
                "phi_angle": int(phi_angles[i]),
                "energy": float(energies[i]),
                "dipole_x": float(dipoles[i, 0]),
                "dipole_y": float(dipoles[i, 1]),
                "dipole_z": float(dipoles[i, 2]),
                "EF_x": float(ef_vec[0]),
                "EF_y": float(ef_vec[1]),
                "EF_z": float(ef_vec[2]),
            }
    
            for atom in range(forces.shape[1]):
                row[f"F{atom}_x"] = float(forces[i, atom, 0])
                row[f"F{atom}_y"] = float(forces[i, atom, 1])
                row[f"F{atom}_z"] = float(forces[i, atom, 2])
    
            flat_records.append(row)
    
    df_flat = pd.DataFrame(flat_records)
    all_dfs.append(df_flat)
```

```python
import pandas as pd


def plot_together(df_flat):

    # columns to reshape
    value_cols = [
        'dipole_x','dipole_y','dipole_z',
        'EF_x','EF_y','EF_z',
        'F0_x','F0_y','F0_z',
        'F1_x','F1_y','F1_z',
        'F2_x','F2_y','F2_z',
        'F3_x','F3_y','F3_z'
    ]
    
    id_cols = ["uuid", "phi_angle", "energy"]
    
    # melt
    df_long = df_flat.melt(
        id_vars=id_cols,
        value_vars=value_cols,
        var_name="variable",
        value_name="value"
    )
    
    # split into (quantity, axis)
    df_long[["quantity", "axis"]] = df_long["variable"].str.rsplit("_", n=1, expand=True)
    
    # pivot to desired layout
    df_semantic = df_long.pivot_table(
        index=["uuid",  "phi_angle", "axis"],
        columns="quantity",
        values="value"
    ).reset_index()
    
    
    import matplotlib.pyplot as plt
    
    axes_list = ["x", "y", "z"]
    quantities = ["dipole", "EF", "F0", "F1", "F2", "F3"]
    
    fig, axs = plt.subplots(
        nrows=len(axes_list),
        ncols=len(quantities),
        figsize=(3 * len(quantities), 3 * len(axes_list)),
        sharex=True
    )
    
    # ensure 2D array even if degenerate
    axs = axs.reshape(len(axes_list), len(quantities))
    
    for i, ax_name in enumerate(axes_list):
        df_ax = df_semantic[df_semantic["axis"] == ax_name]
    
        for j, q in enumerate(quantities):
            ax = axs[i, j]
    
            for run, df_run in df_ax.groupby("uuid"):
                ax.plot(
                    df_run["phi_angle"],
                    df_run[q],
                    label=run,
                    linewidth=1
                )
    
            # titles only on top row
            if i == 0:
                ax.set_title(q)
    
            # y-labels only on first column
            if j == 0:
                ax.set_ylabel(f"{ax_name}-axis")
    
            # x-labels only on bottom row
            if i == len(axes_list) - 1:
                ax.set_xlabel("phi_angle")
    
    # one shared legend
    handles, labels = axs[0, 0].get_legend_handles_labels()
    fig.legend(handles, labels, loc="upper right", fontsize=8)
    
    plt.tight_layout()
    plt.show()
```

```python
import matplotlib.pyplot as plt
from matplotlib.transforms import blended_transform_factory

# Filtered table:
# - kept everything except rows whose assignment contains "dimer/multimer"
EXPERIMENTAL_BANDS = {
    "H2O2": [
        # (3597.0, "ν5, ν1b"),
        (3588, "ν5, ν1"),
        # (3587.5, "N2 complex"),
        (2653, "ν2 + ν6"),
        # (2643.6, "ν2 + ν6"),
        # (1279.9, "N2 complex"),
        # (1277.0, "ν6"),
        (1271, "ν6"),
         (869, "ν3"),
         (490, "ν4"),
    ],
    "HDO2": [
        (3587.3, "OH stretching, possibly multimer"),
        (1350.3, "OH bending"),
        (1342.6, "OH bending"),
        (992.2,  "N2 complex"),
        (980.9,  "OD bending"),
    ],
    "D2O2": [
        (2655.7, "ν5, ν1b"),
        (2651.3, "ν5, ν1"),
        (2645.3, "ν5, ν1"),
        (2647.2, "N2 complex"),
        (1974.2, "ν2 + ν6"),
        (1965.2, "ν2 + ν6"),
        (955.8,  "N2 complex"),
        (951.3,  "ν6"),
    ],
}


def annotate_experimental_bands(
    ax,
    species,
    bands=None,
    line_color="0.4",
    line_style="--",
    line_width=1.0,
    text_color="0.2",
    fontsize=8,
    rotation=90,
    y_top=0.98,
    y_step=0.08,
    n_stagger=3,
    show_frequency=True,
    show_assignment=True,
):
    """
    Annotate a spectrum with experimental band positions.

    Parameters
    ----------
    ax : matplotlib axis
        Axis to annotate.
    species : str
        One of "H2O2", "HDO2", "D2O2".
    bands : dict or None
        Mapping species -> list of (frequency, assignment).
        If None, uses EXPERIMENTAL_BANDS.
    line_color, line_style, line_width : styling for vertical lines
    text_color, fontsize, rotation : styling for labels
    y_top : float
        Top position for labels in axis fraction coordinates.
    y_step : float
        Vertical staggering step in axis fraction coordinates.
    n_stagger : int
        Number of stagger levels to cycle through.
    show_frequency, show_assignment : bool
        Control what appears in the label text.
    """
    if bands is None:
        bands = EXPERIMENTAL_BANDS

    if species not in bands:
        raise ValueError(f"Unknown species '{species}'. Options: {list(bands)}")

    x0, x1 = ax.get_xlim()
    xmin, xmax = min(x0, x1), max(x0, x1)

    # x in data coordinates, y in axis fraction coordinates
    trans = blended_transform_factory(ax.transData, ax.transAxes)

    for i, (freq, assignment) in enumerate(bands[species]):
        # only annotate if inside visible x-range
        if not (xmin <= freq <= xmax):
            continue

        ax.axvline(freq, color=line_color, ls=line_style, lw=line_width, alpha=0.8)

        parts = []
        if show_frequency:
            parts.append(f"{freq:.1f}")
        if show_assignment:
            parts.append(assignment)
        label = "\n".join(parts)

        y = y_top - y_step * (i % n_stagger)

        ax.text(
            freq,
            y,
            label,
            transform=trans,
            rotation=rotation,
            ha="center",
            va="top",
            fontsize=fontsize,
            color=text_color,
            clip_on=False,
        )


import numpy as np
import matplotlib.pyplot as plt
from matplotlib.transforms import blended_transform_factory

P_H2O2_TABLE5 = [
    {"freq": 3812.87, "IR": {"HF": 15.191,  "MP2": 2.533,   "CID": 3.309},
                     "RS": {"HF": 53.349,   "MP2": 22.911,  "CID": 26.009}},
    {"freq": 3810.34, "IR": {"HF": 72.759,  "MP2": 20.018,  "CID": 24.491},
                     "RS": {"HF": -63.938,  "MP2": -28.768, "CID": -31.378}},
    {"freq": 1443.26, "IR": {"HF": 0.490,   "MP2": 0.256,   "CID": 0.299},
                     "RS": {"HF": -17.797,  "MP2": -12.826, "CID": -13.823}},
    {"freq": 1306.96, "IR": {"HF": 118.048, "MP2": 139.448, "CID": 134.127},
                     "RS": {"HF": 8.545,    "MP2": 0.817,   "CID": 1.713}},
    {"freq": 920.51,  "IR": {"HF": 2.594,   "MP2": 1.248,   "CID": 1.665},
                     "RS": {"HF": 3.749,    "MP2": 2.470,   "CID": 2.799}},
    {"freq": 338.53,  "IR": {"HF": 320.104, "MP2": 291.536, "CID": 294.062},
                     "RS": {"HF": 261.392,  "MP2": 250.415, "CID": 251.641}},
]


def annotate_vcd_table(
    ax,
    table,
    method="MP2",
    show_text=True,
    text_fmt="{freq:.0f}",
    pos_color="red",
    neg_color="blue",
    line_style="--",
    base_lw=1.0,
    scale_linewidth=False,
    max_lw=2.5,
    alpha=0.8,
    y_top=0.98,
    y_step=0.08,
    n_stagger=3,
):
    """
    Annotate a VCD spectrum axis using tabulated frequencies and rotatory strengths.

    Parameters
    ----------
    ax : matplotlib axis
    table : list of dicts
        Data like P_H2O2_TABLE5
    method : str
        "HF", "MP2", or "CID"
    show_text : bool
        If True, place frequency labels near the top
    text_fmt : str
        Format string for frequency labels
    pos_color, neg_color : str
        Colors for positive/negative rotatory strengths
    line_style : str
        Style for vertical lines
    base_lw : float
        Default linewidth
    scale_linewidth : bool
        If True, line width scales with |rotatory strength|
    max_lw : float
        Maximum linewidth when scaling
    alpha : float
        Line/text transparency
    y_top, y_step, n_stagger :
        Control vertical staggering of text labels
    """
    x0, x1 = ax.get_xlim()
    xmin, xmax = min(x0, x1), max(x0, x1)

    vals = [abs(row["RS"][method]) for row in table]
    max_rs = max(vals) if vals else 1.0

    trans = blended_transform_factory(ax.transData, ax.transAxes)

    visible_idx = 0
    for row in table:
        freq = row["freq"]
        rs = row["RS"][method]

        if not (xmin <= freq <= xmax):
            continue

        color = pos_color if rs >= 0 else neg_color

        if scale_linewidth:
            lw = base_lw + (max_lw - base_lw) * (abs(rs) / max_rs)
        else:
            lw = base_lw

        ax.axvline(freq, color=color, ls=line_style, lw=lw, alpha=alpha)

        if show_text:
            y = y_top - y_step * (visible_idx % n_stagger)
            label = text_fmt.format(freq=freq, rs=rs)

            ax.text(
                freq,
                y,
                label,
                transform=trans,
                rotation=90,
                ha="center",
                va="top",
                color=color,
                fontsize=8,
                alpha=alpha,
                clip_on=False,
            )
            visible_idx += 1



RAMAN_TAYLOR_CROSS_1956 = {
    "H2O2_liq_30C": [
        (200.0, 25.0, "", ""),
        (525.0, 25.0, "ν4", ""),
        (880.0, 0.5, "ν3(a)", "pol."),
        (1400.0, 5.0, "ν2(a)", "pol."),
        (2815.0, 15.0, "2ν2(A)", ""),
        (3410.0, 8.0, "ν1(a)", "pol."),
    ],
    "H2O2_liq_minus40C": [
        (880.0, 0.5, "ν3(a)", "pol."),
        (1402.0, 5.0, "ν2(a)", "pol."),
        (3364.0, 8.0, "ν1(a)", "pol."),
    ],
    "D2O2_liq_90pct_minus40C": [
        (880.0, 0.5, "ν3(a)", "pol."),
        (1013.0, 5.0, "ν2(a)", "pol."),
        (2038.0, 10.0, "2ν2(A)", ""),
        (2472.0, 7.0, "ν1(a)", "pol."),
    ],
}


from matplotlib.transforms import blended_transform_factory

def annotate_raman_bands(
    ax,
    dataset="H2O2_liq_30C",
    bands=None,
    line_color="tab:blue",
    text_color="tab:blue",
    line_style=":",
    line_width=1.2,
    alpha=0.85,
    fontsize=8,
    show_uncertainty=True,
    show_assignment=True,
    show_polarization=True,
    y_top=0.98,
    y_step=0.08,
    n_stagger=3,
):
    """
    Annotate Raman bands from Taylor & Cross, J. Chem. Phys. 24, 41 (1956).

    Each entry is:
        frequency, uncertainty, assignment, polarization
    """
    if bands is None:
        bands = RAMAN_TAYLOR_CROSS_1956

    if dataset not in bands:
        raise ValueError(f"Unknown Raman dataset: {dataset}. Options: {list(bands)}")

    x0, x1 = ax.get_xlim()
    xmin, xmax = min(x0, x1), max(x0, x1)
    trans = blended_transform_factory(ax.transData, ax.transAxes)

    visible_idx = 0

    for freq, err, assignment, pol in bands[dataset]:
        if not (xmin <= freq <= xmax):
            continue

        ax.axvline(
            freq,
            color=line_color,
            ls=line_style,
            lw=line_width,
            alpha=alpha,
        )

        if show_uncertainty and err is not None:
            ax.axvspan(
                freq - err,
                freq + err,
                color=line_color,
                alpha=0.08,
                lw=0,
            )

        label_parts = [f"{freq:.0f}"]

        if show_assignment and assignment:
            label_parts.append(assignment)

        if show_polarization and pol:
            label_parts.append(pol)

        label = "\n".join(label_parts)
        y = y_top - y_step * (visible_idx % n_stagger)

        ax.text(
            freq,
            y,
            label,
            transform=trans,
            rotation=90,
            ha="center",
            va="top",
            fontsize=fontsize,
            color=text_color,
            alpha=alpha,
            clip_on=False,
        )

        visible_idx += 1
```

```python
import numpy as np
import matplotlib.pyplot as plt
from ase import io

from mmml.models.EF.ase_calc_EF import AseCalculatorEF
from mmml.models.EF.calc_spectra import (
    compute_normal_modes, compute_ir, compute_raman, compute_vcd
)

# --------------------------------------------------
# Input structure + calculator
# --------------------------------------------------
atoms = io.read("hper_b.xyz")

calc = AseCalculatorEF(
    runs[run_index]["params"],
    config_path=runs[run_index]["config"],
    electric_field=[0, 0, 0],
    field_scale=100,
)
atoms.calc = calc

# Example: coordinates to scan over
frames = dih_data["R"]   # shape: (n_frames, n_atoms, 3)

# --------------------------------------------------
# Helpers
# --------------------------------------------------
def lorentzian(x, x0, gamma):
    return gamma / (np.pi * ((x - x0)**2 + gamma**2))

def broaden_spectrum(freq_axis, mode_freqs, mode_intensities, gamma=10.0, min_freq=20.0):
    y = np.zeros_like(freq_axis, dtype=float)
    for f, I in zip(mode_freqs, mode_intensities):
        if np.isfinite(f) and f > min_freq:
            y += I * lorentzian(freq_axis, f, gamma)
    return y

def safe_norm(y):
    m = np.max(np.abs(y))
    return y / m if m > 0 else y

# --------------------------------------------------
# Frequency axis
# --------------------------------------------------
fmin, fmax, npts = 50.0, 4000.0, 4000
freq_axis = np.linspace(fmin, fmax, npts)

# --------------------------------------------------
# Compute spectra for all frames
# --------------------------------------------------
ir_all = []
raman_all = []
vcd_all = []

for R in frames:
    atoms.set_positions(R)

    H = calc.get_hessian(atoms)
    APT = calc.get_atomic_polar_tensor(atoms)
    freq, evec_mw, evec_cart = compute_normal_modes(H, atoms.get_masses())

    ir_int, _ = compute_ir(APT, evec_cart)
    raman_act, _ = compute_raman(calc, atoms, evec_cart, freq)
    AAT, _ = calc.get_aat_born(atoms)
    vcd_rot, _, _ = compute_vcd(APT, AAT, evec_cart)

    ir_spec = broaden_spectrum(freq_axis, freq, ir_int, gamma=12.0)
    raman_spec = broaden_spectrum(freq_axis, freq, raman_act, gamma=12.0)
    vcd_spec = broaden_spectrum(freq_axis, freq, vcd_rot, gamma=12.0)

    ir_all.append(safe_norm(ir_spec))
    raman_all.append(safe_norm(raman_spec))
    vcd_all.append(safe_norm(vcd_spec))

ir_all = np.asarray(ir_all)
raman_all = np.asarray(raman_all)
vcd_all = np.asarray(vcd_all)
```

```python

# --------------------------------------------------
# Plotting: same style as original, but shifted
# --------------------------------------------------
def plot_shifted_panel(ax, x, spectra, offset=1.2, scale=0.8,
                       line_color="black", fill_color=None, alpha=0.25,
                       ylabel=""):
    """
    Plot normalized spectra with vertical shifts.
    """
    n = spectra.shape[0]

    for i in range(n):
        y0 = i * offset
        y = y0 + scale * spectra[i] 

        ax.plot(x, y, lw=1.4, color=line_color)

        if fill_color is not None:
            ax.fill_between(x, y0, y, color=fill_color, alpha=alpha)

    ax.set_ylabel(ylabel)
    ax.set_yticks([])

def plot_shifted_vcd_panel(ax, x, spectra, offset=1.2, scale=0.6,
                           line_color="k", pos_color="red", neg_color="blue",
                           alpha=0.35, ylabel="VCD (norm.)"):
    """
    Plot shifted VCD spectra with positive = red and negative = blue.
    """
    n = spectra.shape[0]

    for i in range(n):
        y0 = i * offset
        s = scale * spectra[i]
        y = y0 + s

        ax.plot(x, y, lw=1.2, color=line_color)

        ax.fill_between(x, y0, y, where=(s >= 0), color=pos_color, alpha=alpha)
        ax.fill_between(x, y0, y, where=(s < 0), color=neg_color, alpha=alpha)

    ax.set_ylabel(ylabel)
    ax.set_yticks([])

def add_angle_axis(ax, angles, offset, tick_step=4, label="Dihedral angle (deg)"):
    ax_r = ax.twinx()
    ax_r.set_ylim(ax.get_ylim())

    positions = np.arange(len(angles)) * offset
    ax_r.set_yticks(positions[::tick_step])
    ax_r.set_yticklabels([f"{a:d}" for a in angles[::tick_step]])
    ax_r.set_ylabel(label)

    # ax_r.axhline(0, color="k", alpha=0.5)
    # ax_r.axhline(-180, color="k", alpha=0.5)
    # ax_r.axhline(180, color="k", alpha=0.5)

    return ax_r
```

```python
# --------------------------------------------------
# Figure
# --------------------------------------------------
fig, axes = plt.subplots(3, 1, figsize=(10, 10), sharex=True)
SCALE = 7
plot_shifted_panel(
    axes[0], freq_axis, ir_all,
    offset=1.15, scale=SCALE,
    line_color="black",
    fill_color="gray",
    alpha=0.25,
    ylabel="IR (norm.)"
)
axes[0].set_title("IR / Raman / VCD (harmonic)")

plot_shifted_panel(
    axes[1], freq_axis, raman_all,
    offset=1.15, scale=SCALE,
    line_color="tab:blue",
    fill_color="tab:blue",
    alpha=0.25,
    ylabel="Raman (norm.)"
)

plot_shifted_vcd_panel(
    axes[2], freq_axis, vcd_all,
    offset=1.15, scale=SCALE * 1,
    line_color="k",
    pos_color="red",
    neg_color="blue",
    alpha=0.35,
    ylabel="VCD (norm.)"
)

axes[2].set_xlabel(r"Wavenumber (cm$^{-1}$)")

# optional, like your original
for ax in axes:
    ax.grid(alpha=0.2)
    ax.set_xlim(800, 2000)

offset = 1.15
# add right-hand angle axis
add_angle_axis(axes[0], phi_angles, offset, tick_step=4, label="Angle (deg)")
add_angle_axis(axes[1], phi_angles, offset, tick_step=4, label="Angle (deg)")
add_angle_axis(axes[2], phi_angles, offset, tick_step=4, label="Angle (deg)")


plt.tight_layout()
plt.savefig("h2o2_uknownpleasures.png", bbox_inches="tight", dpi=300)
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib as mpl
from pathlib import Path


HARTREE_TO_EV = 27.2114
DIPOLE_AU_TO_DEBYE = 2.541746  # 1 e a0 = 2.541746 D
DEBYE_TO_E_ANG = 1.0 / 4.80320427
FORCE_AU_TO_EV_PER_ANG = 51.422067
from cmap import Colormap

def get_model_style(row, max_degree_max=5, cmap_name="viridis"):
    """
    Shared style rule:
      - color: max_degree
      - linestyle: pseudotensor flag
      - marker: max_degree, to improve distinguishability
    """
    ell = int(row["model.max_degree"])
    pt = bool(row["model.include_pseudotensors"])

    # cmap = mpl.colormaps[cmap_name]
    cmap = Colormap('batlow_r').to_mpl()
    color = cmap(ell / max_degree_max)

    linestyles = {
        False: "-",
        True: (0, (4, 2)),   # custom dashed
    }

    markers = ["o", "s", "^", "D", "v", "P", "X", "*"]
    marker = markers[ell % len(markers)]

    label = rf"$\ell={ell}$, pt={pt}"

    return {
        "color": color,
        "linestyle": linestyles[pt],
        "marker": marker,
        "label": label,
    }


def load_truth_npz(path, field_label=None):
    """
    Load one ground-truth .npz and return both raw arrays and a flat dataframe.

    Expected keys:
      E, F, Dxyz, optionally Ef
    """
    data = np.load(path)

    E = np.asarray(data["E"])
    F = np.asarray(data["F"])
    D = np.asarray(data["Dxyz"])

    n_scan = len(E)

    if "Ef" in data.files:
        Ef = np.asarray(data["Ef"])
    else:
        Ef = None

    rows = []
    for i in range(n_scan):
        row = {
            "field_label": field_label,
            "scan_index": i,
            "energy_true": float(E[i]),
            "dipole_x_true": float(D[i, 0]),
            "dipole_y_true": float(D[i, 1]),
            "dipole_z_true": float(D[i, 2]),
        }

        for atom in range(F.shape[1]):
            for j, comp in enumerate(["x", "y", "z"]):
                row[f"F{atom}_{comp}_true"] = float(F[i, atom, j])

        rows.append(row)

    df_true = pd.DataFrame(rows)

    return {
        "raw": data,
        "E": E,
        "F": F,
        "Dxyz": D,
        "Ef": Ef,
        "df": df_true,
    }


def prepare_model_df(df_model, models):
    """
    Attach model metadata to one model scan dataframe.
    """
    needed = ["uuid", "model.max_degree", "model.include_pseudotensors"]
    model_info = models[needed].drop_duplicates("uuid")

    df = df_model.copy()
    df = df.merge(model_info, on="uuid", how="left")

    df = df.sort_values(["uuid", "phi_angle"]).reset_index(drop=True)
    df["scan_index"] = df.groupby("uuid").cumcount()

    return df


def iter_models(df_model_meta, max_degree_values=range(6)):
    """
    Yield UUID-specific model dataframes in a stable plotting order.
    """
    for ell in max_degree_values:
        sub_ell = df_model_meta[df_model_meta["model.max_degree"] == ell]

        sub_ell = sub_ell.sort_values(
            ["model.include_pseudotensors", "uuid"]
        )

        for uuid, sub_uuid in sub_ell.groupby("uuid", sort=False):
            meta = sub_uuid.iloc[0]
            yield uuid, sub_uuid.sort_values("scan_index"), meta


def collect_legend_entries(axes):
    """
    Collect and de-duplicate legend entries from one or more axes.
    """
    handles = []
    labels = []

    for ax in np.ravel(axes):
        h, l = ax.get_legend_handles_labels()
        handles.extend(h)
        labels.extend(l)

    unique = {}
    for h, l in zip(handles, labels):
        if l not in unique:
            unique[l] = h

    return list(unique.values()), list(unique.keys())


def make_shared_legend_figure(handles, labels, ncol=4, figsize=(8, 1.2), save_path=None):
    """
    Create a standalone legend-only figure.
    """
    fig, ax = plt.subplots(figsize=figsize)
    ax.axis("off")

    ax.legend(
        handles,
        labels,
        loc="center",
        ncol=ncol,
        frameon=False,
        handlelength=2.8,
        columnspacing=1.5,
        handletextpad=0.6,
    )

    fig.tight_layout()

    if save_path is not None:
        fig.savefig(save_path, dpi=300, bbox_inches="tight")

    return fig, ax


def plot_relative_energy(
    df_model,
    truth,
    models,
    title=None,
    save_path=None,
    max_degree_values=range(6),
):
    """
    Plot relative energy scan:
      truth: (E - Emin) * 27.2114
      model: energy - Emin

    Note: this assumes model energies are already in the desired unit.
    If model energy is Hartree, multiply it by HARTREE_TO_EV too.
    """
    df = prepare_model_df(df_model, models)

    E_true_rel = (truth["E"] - truth["E"].min()) * HARTREE_TO_EV

    fig, ax = plt.subplots(figsize=(7, 5))

    plotted_truth = False

    for uuid, sub, meta in iter_models(df, max_degree_values=max_degree_values):
        style = get_model_style(meta)

        phi = sub["phi_angle"].to_numpy()
        E_model_rel = sub["energy"].to_numpy() - sub["energy"].min()

        if not plotted_truth:
            ax.plot(phi, E_true_rel, color="k", linewidth=2.5, label="truth")
            plotted_truth = True

        ax.plot(
            phi,
            E_model_rel,
            color=style["color"],
            linestyle=style["linestyle"],
            marker=style["marker"],
            markevery=max(1, len(phi) // 12),
            markersize=4,
            linewidth=1.6,
            alpha=0.80,
            label=style["label"],
        )

    ax.set_xlabel(r"$\phi$")
    ax.set_ylabel(r"$\Delta E$")
    if title is not None:
        ax.set_title(title)

    # add_shared_legend(fig, [ax])
    fig.tight_layout(rect=[0, 0, 1, 0.88])

    if save_path is not None:
        fig.savefig(save_path, dpi=300, bbox_inches="tight")

    return fig, ax


def plot_dipoles(
    df_model,
    truth,
    models,
    title=None,
    save_path=None,
    truth_unit="e_ang",
    model_unit="e_ang",
    max_degree_values=range(6),
):
    """
    Plot D_x, D_y, D_z on a shared figure.

    truth_unit options:
      - "e_ang": divide truth Dxyz by 4.80320427
      - "debye": leave truth as Debye if already Debye
      - "au": convert atomic units to Debye

    model_unit currently only affects ylabel text; no conversion is applied to model.
    """
    df = prepare_model_df(df_model, models)

    D_true = np.asarray(truth["Dxyz"])

    if truth_unit == "e_ang":
        D_true_plot = D_true * DEBYE_TO_E_ANG
        ylabel = r"Dipole / $e\,\AA$"
    elif truth_unit == "debye":
        D_true_plot = D_true
        ylabel = "Dipole / Debye"
    elif truth_unit == "au":
        D_true_plot = D_true * DIPOLE_AU_TO_DEBYE
        ylabel = "Dipole / Debye"
    else:
        raise ValueError(f"Unknown truth_unit: {truth_unit}")

    comps = ["x", "y", "z"]

    fig, axes = plt.subplots(
        3,
        1,
        figsize=(7, 8),
        sharex=True,
        sharey=False,
    )

    for c_idx, comp in enumerate(comps):
        ax = axes[c_idx]
        plotted_truth = False

        for uuid, sub, meta in iter_models(df, max_degree_values=max_degree_values):
            style = get_model_style(meta)

            phi = sub["phi_angle"].to_numpy()

            if not plotted_truth:
                ax.plot(
                    phi,
                    D_true_plot[:, c_idx],
                    color="k",
                    linewidth=2.5,
                    label="truth",
                )
                plotted_truth = True

            ax.plot(
                phi,
                sub[f"dipole_{comp}"].to_numpy(),
                color=style["color"],
                linestyle=style["linestyle"],
                marker=style["marker"],
                markevery=max(1, len(phi) // 12),
                markersize=4,
                linewidth=1.5,
                alpha=0.80,
                label=style["label"],
            )

        ax.set_ylabel(rf"$D_{comp}$")
        ax.axhline(0.0, color="0.8", linewidth=0.8)

    axes[-1].set_xlabel(r"$\phi$")
    fig.supylabel(ylabel)

    if title is not None:
        fig.suptitle(title, y=1.05)

    # add_shared_legend(fig, axes)
    fig.tight_layout(rect=[0, 0, 1, 0.90])

    if save_path is not None:
        fig.savefig(save_path, dpi=300, bbox_inches="tight")

    return fig, axes


def plot_forces(
    df_model,
    truth,
    models,
    atoms=None,
    components=("x", "y", "z"),
    title=None,
    save_path=None,
    truth_scale=FORCE_AU_TO_EV_PER_ANG,
    max_degree_values=range(6),
):
    """
    Plot force components on a shared figure.

    Default: all atom/component combinations in a grid.

    truth_scale:
      use 51.422067 if truth forces are Hartree/Bohr and model forces are eV/Angstrom.
      use 1.0 if both are already in the same units.
    """
    df = prepare_model_df(df_model, models)

    F_true = np.asarray(truth["F"]) * truth_scale

    n_atoms = F_true.shape[1]
    if atoms is None:
        atoms = list(range(n_atoms))

    comp_to_idx = {"x": 0, "y": 1, "z": 2}

    nrows = len(atoms)
    ncols = len(components)

    fig, axes = plt.subplots(
        nrows,
        ncols,
        figsize=(4.0 * ncols, 2.4 * nrows),
        sharex=True,
        squeeze=False,
    )

    for row_idx, atom in enumerate(atoms):
        for col_idx, comp in enumerate(components):
            ax = axes[row_idx, col_idx]
            comp_idx = comp_to_idx[comp]

            plotted_truth = False

            for uuid, sub, meta in iter_models(df, max_degree_values=max_degree_values):
                style = get_model_style(meta)

                phi = sub["phi_angle"].to_numpy()

                if not plotted_truth:
                    ax.plot(
                        phi,
                        F_true[:, atom, comp_idx],
                        color="k",
                        linewidth=2.2,
                        label="truth",
                    )
                    plotted_truth = True

            ax.plot(
                phi,
                sub[f"F{atom}_{comp}"].to_numpy(),
                color=style["color"],
                linestyle=style["linestyle"],
                marker=style["marker"],
                markevery=max(1, len(phi) // 12),
                markersize=3.5,
                linewidth=1.4,
                alpha=0.75,
                label=style["label"],
            )

            ax.axhline(0.0, color="0.85", linewidth=0.8)
            ax.set_title(rf"$F_{{{atom},{comp}}}$")

            if row_idx == nrows - 1:
                ax.set_xlabel(r"$\phi$")

            if col_idx == 0:
                ax.set_ylabel("Force")

    if title is not None:
        fig.suptitle(title, y=1.04)

    # add_shared_legend(fig, axes, ncol=4)
    fig.tight_layout(rect=[0, 0, 1, 0.90])

    if save_path is not None:
        fig.savefig(save_path, dpi=300, bbox_inches="tight")

    return fig, axes
```

```python
indices = np.argwhere((dihedrals < 50) | (dihedrals > 310))
```

```python
import numpy as np

# traj: ASE Trajectory, or list/tuple of ASE Atoms from traj[i]
# indices: list/array of frame indices, e.g. [0, 10, 20, 50]
out_path = "selected_frames.npz"

R_list = []
N_list = []
Z_ref = trajs[0].get_atomic_numbers()

for i in indices:
    atoms = trajs[int(i[0])]  # use traj[int(i)] if indices are floats
    R = np.asarray(atoms.get_positions(), dtype=np.float64)
    Z = np.asarray(atoms.get_atomic_numbers(), dtype=np.int32)
    if Z_ref is None:
        Z_ref = Z
    elif not np.array_equal(Z_ref, Z):
        raise ValueError(f"Frame {i}: atomic numbers differ from first selected frame")
    R_list.append(R)
    N_list.append(len(atoms))

R_arr = np.stack(R_list, axis=0)  # (n_frames, n_atoms, 3)
N_arr = np.asarray(N_list, dtype=np.int32)  # (n_frames,)

np.savez_compressed(out_path, R=R_arr, Z=Z_ref, N=N_arr)
print("Saved", out_path, "R", R_arr.shape, "Z", Z_ref.shape, "N", N_arr.shape)
```
