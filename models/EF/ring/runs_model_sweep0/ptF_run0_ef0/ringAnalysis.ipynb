```python
import numpy as np

# Source - https://stackoverflow.com/a/56515845
# Posted by sgt pepper
# Retrieved 2026-05-19, License - CC BY-SA 4.0

def unit_vector(vector):
    """ Returns the unit vector of the vector"""
    return vector / np.linalg.norm(vector)

def angle_between(vector1, vector2):
    """ Returns the angle in radians between given vectors"""
    v1_u = unit_vector(vector1)
    v2_u = unit_vector(vector2)
    minor = np.linalg.det(
        np.stack((v1_u[-2:], v2_u[-2:]))
    )
    if minor == 0:
        raise NotImplementedError('Too odd vectors =(')
    return np.sign(minor) * np.arccos(np.clip(np.dot(v1_u, v2_u), -1.0, 1.0))
```

```python
def iso_comp(df, ax, ylab=False, xlab=False):
    # plt.figure(figsize=(7, 4))
    ax.hist(df["cos_theta_to_lab_z"], bins=30, density=True, alpha=0.26, label="Observed", color="k")
    ax.axhline(0.5, linestyle="--", label="Uniform isotropic reference", c="k")
    if xlab:
        ax.set_xlabel(r"$\cos(\theta)$")
    if ylab:
        ax.set_ylabel(r"$P(\cos(\theta))$")
    # plt.title(r"Reference check: $\cos(\theta)$ uniform on [-1, 1]")
    # plt.legend()
    # plt.tight_layout()
    # plt.close()
    ax.set_ylim(0, 5)
    ax.set_xlim(-1.0, 1.0)
    return ax


def cp_comp(df, ax, ylab=False, xlab=False):
    ax.hist2d(df["x_cp"], df["y_cp"], bins=30, vmin=0, vmax=30, cmap="plasma")
    # plt.colorbar(label="Count")
    if xlab:
        ax.set_xlabel("q cos(P) / Å")
    if ylab:
        ax.set_ylabel("q sin(P) / Å")
    ax.set_xlim(-.5, .5)
    ax.set_xlim(-.5, .5)
    return ax
```

```python
fig = plt.figure(figsize=(10, 5))

gs = fig.add_gridspec(
    2, 5,
    width_ratios=[1, 1, 0.25, 1, 1],  # middle spacer
    wspace=0.5,
    hspace=0.35,
)

axs = np.array([
    [fig.add_subplot(gs[0, 0]), fig.add_subplot(gs[0, 1]),
     fig.add_subplot(gs[0, 3]), fig.add_subplot(gs[0, 4])],
    [fig.add_subplot(gs[1, 0]), fig.add_subplot(gs[1, 1]),
     fig.add_subplot(gs[1, 3]), fig.add_subplot(gs[1, 4])],
])

k = keys[0]
print(k)
df = results[k]
ax1 = iso_comp(df, axs[0,0], ylab=True)
cp_ax1 = cp_comp(df, axs[0,2], ylab=True)
k = keys[1]
print(k)
df = results[k]
ax2 = iso_comp(df, axs[0,1])
cp_ax2 = cp_comp(df, axs[0,3])
k = keys[-2]
print(k)
df = results[k]
ax3 = iso_comp(df, axs[1,0], ylab=True, xlab=True)
cp_ax3 = cp_comp(df, axs[1,2], ylab=True, xlab=True)
k = keys[-1]
print(k)
df = results[k]
ax4 = iso_comp(df, axs[1,1], xlab=True)
cp_ax4 = cp_comp(df, axs[1,3], xlab=True)


for label, ax in zip("abefcdgh", axs.ravel()):
    ax.text(
        -0.32, 1.15, "("+label+")",
        transform=ax.transAxes,
        ha="left",
        va="top",
        fontsize=12,
        # fontweight="bold",
        bbox=dict(facecolor="white", edgecolor="none", alpha=0.7, pad=1.5),
    )


plt.savefig("ring_analysis.pdf", bbox_inches="tight")
```
